```python
import matplotlib.pyplot as plt

# Ensure 'Order Date' is datetime and set as index
df['Order Date'] = pd.to_datetime(df['Order Date'])
df_daily_sales = df.set_index('Order Date')['Sales'].resample('D').sum()

# Calculate 7-day rolling mean
df_rolling_mean = df_daily_sales.rolling(window=7).mean()

# Plot original daily sales and rolling mean
plt.figure(figsize=(14, 7))
plt.plot(df_daily_sales, label='Daily Sales', alpha=0.8)
plt.plot(df_rolling_mean, label='7-Day Rolling Mean', color='red')
plt.title('Daily Sales and 7-Day Rolling Mean')
plt.xlabel('Date')
plt.ylabel('Sales')
plt.legend()
plt.grid(True)
plt.show()
```

TASK-3 : CALCULATE PERIOD-OVER-PERIOD DAILY GROWTH USING .SHIFT()

```python
df_daily_sales = df.set_index('Order Date')['Sales'].resample('D').sum()

# Create a lag-1 column 'Sales_Prev_Day'
df_daily_sales_with_lag = pd.DataFrame(df_daily_sales)
df_daily_sales_with_lag['Sales_Prev_Day'] = df_daily_sales_with_lag['Sales'].shift(1)

# Calculate daily percentage change in sales
df_daily_sales_with_lag['Daily_Sales_Change_Pct'] = ((df_daily_sales_with_lag['Sales'] - df_daily_sales_with_lag['Sales_Prev_Day']) / df_daily_sales_with_lag['Sales_Prev_Day']) * 100

display(df_daily_sales_with_lag.head())
```

TASK-2 : MATPLOTLIB 3D-SURFACE PLOT OF SYNTHETIC SALES-PROFIT SURFACE

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# 1. Prepare data ranges based on the dataframe
x_range = np.linspace(df['Sales'].min(), df['Sales'].max(), 50)
y_range = np.linspace(df['Profit'].min(), df['Profit'].max(), 50)

# 2. Create the meshgrid
X, Y = np.meshgrid(x_range, y_range)

# 3. Create a synthetic smooth Z surface (e.g., a Gaussian or wave function)
# This represents a hypothetical 'smooth' Quantity distribution
Z = 10 * np.sin(np.sqrt(X**2 + Y**2) / 1000)

# 4. Initialize Plot
fig = plt.figure(figsize=(12, 8))
ax = fig.add_subplot(111, projection='3d')

# 5. Plot the surface
surf = ax.plot_surface(X, Y, Z, cmap='viridis', edgecolor='none', alpha=0.9)

# 6. Customize axes
ax.set_xlabel('Sales')
ax.set_ylabel('Profit')
ax.set_zlabel('Quantity (Synthetic)')
ax.set_title('Smooth 3D Surface Plot')

# 7. Add colorbar and show
fig.colorbar(surf, ax=ax, shrink=0.5, aspect=10)
plt.show()
```

TASK - 1 : MISSING DATA MATRIX PATTERN USING MISSINGO

```python
import numpy as np
import pandas as pd
import missingno as msno
import matplotlib.pyplot as plt

# 1. Load data & take a subset of 200 rows and specific columns
cols = ['Sales', 'Profit', 'Quantity', 'Discount', 'Category', 'Sub-Category', 'Region']
df_subset = df[cols].head(200).copy()

# 2. Inject artificial missing values (as set up in Cell 1 with np.random.seed(42))
np.random.seed(42)

# Add missingness to numerical columns (~30% missing)
for col in ['Sales', 'Profit', 'Discount']:
    df_subset.loc[np.random.rand(len(df_subset)) < 0.3, col] = np.nan

# Add sparse missingness to Category (~1% missing)
df_subset.loc[np.random.rand(len(df_subset)) < 0.01, 'Category'] = np.nan

# 3. Plot the missingness matrix
msno.matrix(df_subset)
plt.title('Missing Data Matrix of Superstore Dataset')
plt.show()
```

```python
from dash import Dash, dcc, html, Input, Output
import plotly.express as px

# Initialize the Dash app
app = Dash(__name__)

# Get unique regions for the dropdown
regions = df['Region'].unique()
regions.sort()

# Define the app layout
app.layout = html.Div([
    html.H1("Superstore Sales Dashboard: Region Selector"),

    html.Div([
        html.Label("Select Region:"),
        dcc.Dropdown(
            id='region-dropdown',
            options=[{'label': region, 'value': region} for region in regions],
            value=regions[0],  # Default selected value
            clearable=False
        )
    ]),

    html.Div(id='output-container') # Output will be displayed here
])

# Define the callback to update the output based on dropdown selection
@app.callback(
    Output('output-container', 'children'),
    Input('region-dropdown', 'value')
)
def update_output(selected_region):
    filtered_df = df[df['Region'] == selected_region]

    # Display some summary stats for the selected region as a starting point
    return html.Div([
        html.H3(f"Data for Region: {selected_region}"),
        html.P(f"Total Sales: ${filtered_df['Sales'].sum():,.2f}"),
        html.P(f"Total Profit: ${filtered_df['Profit'].sum():,.2f}"),
        html.P(f"Number of Orders: {len(filtered_df)}"),
        html.H4("First 5 rows of filtered data:"),
        html.Table(
            [html.Tr([html.Th(col) for col in filtered_df.columns])] +
            [html.Tr([html.Td(filtered_df.iloc[i][col]) for col in filtered_df.columns]) for i in range(min(len(filtered_df), 5))]
        )
    ])

# To run the app in Colab, you need to use a specific port and make it public
# If you're running this in a local environment, you can just use app.run_server(debug=True)

# Note: This will block the notebook cell. You might need to stop the cell execution
# to continue with other tasks in the notebook after viewing the dashboard.
print("Dash app defined. To run it, execute the next cell.")
```

Task 9: Dynamic Discount Histogram (Plotly Graph).

```python
from dash import Dash, dcc, html, Input, Output
import plotly.express as px

# Initialize the Dash app for the Discount Histogram. Using a distinct app variable and port.
app_discount_histogram = Dash(__name__)

# Define the app layout
app_discount_histogram.layout = html.Div([
    html.H1("Dynamic Discount Histogram"),
    html.Div([
        html.Label("Select Number of Bins:"),
        dcc.Slider(
            id='bin-slider',
            min=1,
            max=50,
            step=1,
            value=20, # Default number of bins
            marks={i: str(i) for i in range(0, 51, 5)}
        )
    ]),
    dcc.Graph(id='discount-histogram-chart')
])

# Define the callback to update the histogram based on slider selection
@app_discount_histogram.callback(
    Output('discount-histogram-chart', 'figure'),
    Input('bin-slider', 'value')
)
def update_discount_histogram(n_bins):
    fig = px.histogram(df, x='Discount', nbins=n_bins,
                     title=f'Distribution of Discount (with {n_bins} bins)',
                     labels={'Discount': 'Discount Amount', 'count': 'Number of Orders'},
                     marginal='rug', # Add a rug plot to show individual data points
                     template='plotly_white')

    fig.update_layout(bargap=0.1) # Add some gap between bars for better visualization
    return fig

print("Dash app for Dynamic Discount Histogram defined. Execute the next cell to run it.")
```
